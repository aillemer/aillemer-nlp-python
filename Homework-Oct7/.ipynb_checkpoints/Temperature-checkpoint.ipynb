```python
def FtoC(a, b): #Задаём функции по переводу из фаренгейтов в градусы цельсия и наоборот
    a = (5 / 9) * (a - 32)
    b = "C"
    t = str(a) + b
    return t


def CtoF(a, b):
    a = (5 / 9) * a + 32
    b = "F"
    t = str(a) + b
    return t


def mysplit(x): #Основная функция которая делит входную строку на 3: в первой будет находится минус (если он есть), во второй - число, в третьей - символ единицы измерения
    strt = x.strip("0123456789AaBbCcDdEeFfGgHhIiJjKkLlMmNnOoPpQqRrSsTtUuVvWwXxYyZzАаБбВвГгДдЕеЁёЖжЗзИиЙйКкЛлМмНнОоПпРрСсТтУуФфХхЦцЧчШшЩщЪъЫыЬьЭэЮюЯя")
    mdl = x.strip("-AaBbCcDdEeFfGgHhIiJjKkLlMmNnOoPpQqRrSsTtUuVvWwXxYyZzАаБбВвГгДдЕеЁёЖжЗзИиЙйКкЛлМмНнОоПпРрСсТтУуФфХхЦцЧчШшЩщЪъЫыЬьЭэЮюЯя")
    ed = x.strip("-0123456789")
    b = ed
    a = strt + mdl
    a = int(a)
    if b == "C":
        print(CtoF(a, b))
    elif b == "F":
        print(FtoC(a, b))
    else:
        print("Неверная градусная мера.")


a = input("Введите температуру в градусах Цельсия или Фаренгейта (пр. 140F):")
mysplit(a)
```
