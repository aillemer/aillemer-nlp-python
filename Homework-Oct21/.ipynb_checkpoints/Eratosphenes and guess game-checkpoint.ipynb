## *Задание №1* (нахождение простых чисел с помощью решета Эратосфена):

```python
def numchck(x):
    try:  # проверяем полученную строку на возможность перевода в целочисленный тип данных
        int(x)
    except ValueError:
        print("Введите число!")
        exit(0)
    else:
        x = int(x)
    if x <= 2:
        print("Ошибка: вы ввели значение меньше минимума.")
        exit(0)


def eratosthenes(x):
    resheto = list(range(x + 1))  # строим список от 0 до x (a.k.a. числа, до которого мы будем искать простые числа)
    resheto[1] = (0) # убираем единицу (которая не считается простым числом)
    for i in resheto:
        if i > 1:
            for j in range(2 * i, len(resheto), i):
                resheto[j] = 0
    resheto.sort()  # сортируем список, чтобы все нули оказались в начале
    while resheto[0] == 0:
        resheto.pop(resheto[0])  # удаляем все нули из списка
    return "\n".join(map(str, resheto)) # переводим все содержимое списка в строки и добавляем \n чтобы каждое число выводилось на новой строке


N = input("Введите число, до которого вы хотите найти простые числа: ")
numchck(N)
N = int(N)
print(eratosthenes(N))
```

## *Задание №2* (игра на угадывание числа):

```python
from random import randint

a = randint(0, 99) # используем randint для генерации псевдослучайного целого числа

b = input("Загадано некое число а от 0 до 99. Угадайте его: ")
try: # проверка на перевод в целочисленный ТД
    int(b)
except ValueError:
    print("Введите число!")
    exit(0)
else:
    b = int(b)
    if b < 0:
        print("Ошибка: вы ввели значение меньше 0.")
        exit(0)
    elif b > 99:
        print("Ошибка: вы ввели значение больше 99.")
        exit(0)

    while b != a: # используем while чтобы игра не заканчивалась пока игрок не угадает число
        if b < a:
            b = int(
                input(
                    "Не угадали! Загаданное число больше вашего, попробуйте ещё раз: "
                )
            )
        if b > a:
            b = int(
                input(
                    "Не угадали! Загаданное число меньше вашего, попробуйте ещё раз: "
                )
            )
        if b == a:
            print("Вы угадали!")
```

#### *Пример работы программы:*

Не угадали! Загаданное число меньше вашего, попробуйте ещё раз:  10<br>
Не угадали! Загаданное число больше вашего, попробуйте ещё раз:  12<br>
Вы угадали!
